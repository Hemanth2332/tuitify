```python
import numpy as np
from sentence_transformers import SentenceTransformer
from light_embed import TextEmbedding # or the relevant light-embed class

text = "This is a test sentence to check embedding consistency."

# 1. Old Model
st_model = SentenceTransformer("all-MiniLM-L6-v2")
vec_st = st_model.encode(text)

# 2. New Model
le_model = TextEmbedding("sentence-transformers/all-MiniLM-L6-v2") # adjust to light-embed's exact syntax
vec_le = le_model.encode(text)

# Flatten if necessary
vec_st = np.array(vec_st).flatten()
vec_le = np.array(vec_le).flatten()

# Check norm
print("ST Vector Norm:", np.linalg.norm(vec_st))
print("LE Vector Norm:", np.linalg.norm(vec_le))

# Cosine similarity
cosine_sim = np.dot(vec_st, vec_le) / (np.linalg.norm(vec_st) * np.linalg.norm(vec_le))
print(f"Cosine Similarity: {cosine_sim:.6f}")
```

```python
import numpy as np

# Test with long text and mixed batch
batch_texts = [
    "Short query",
    "A much longer query description with a lot more tokens to test how the tokenizer handles truncation and padding across different sequence lengths in a single batch call.",
    "Another medium length test string."
]

# ST Batch
vecs_st = st_model.encode(batch_texts)

# LE Batch
vecs_le = le_model.encode(batch_texts)

for i, text in enumerate(batch_texts):
    sim = np.dot(vecs_st[i], vecs_le[i]) / (np.linalg.norm(vecs_st[i]) * np.linalg.norm(vecs_le[i]))
    print(f"Item {i} Cosine Similarity: {sim:.6f}")
```
